```python
#account_user_contacts
import pandas as pd
import ast

# 1. 데이터 로드 (경로는 작업 환경에 맞게 수정)
user_contacts = pd.read_csv("votes/accounts_user_contacts.csv")
print(f"▶ 원본 데이터 건수: {len(user_contacts):,}건")

# 2. 방어적 중복 제거
user_contacts = user_contacts.drop_duplicates(subset=['user_id'], keep='last')

# 3. 이상치 정제 (연락처 수가 0인 데이터) - 필요 시 주석 해제
# user_contacts = user_contacts[user_contacts['contacts_count'] > 0]

# 4. 문자열 리스트를 실제 리스트로 안전하게 변환하는 함수 (에러 방지 완벽 적용)
def safe_literal_eval_contacts(val):
    if isinstance(val, list):
        return val
    if not isinstance(val, str):
        return []
    try:
        parsed = ast.literal_eval(val)
        if isinstance(parsed, list):
            return parsed
        return []
    except (ValueError, SyntaxError):
        return []

# invite_user_id_list 컬럼 형태 변환 (문자열 -> 리스트)
user_contacts['invite_user_id_list'] = user_contacts['invite_user_id_list'].apply(safe_literal_eval_contacts)

# 💡 5. 초대한 유저 수(invite_count) 파생변수 생성
# 4번에서 이미 완벽한 리스트 형태로 변환해 두었기 때문에, 단순히 len 함수만 적용하면 됩니다!
user_contacts['invite_count'] = user_contacts['invite_user_id_list'].apply(len)

# 인덱스 초기화
user_contacts = user_contacts.reset_index(drop=True)

print(f"🎉 전처리 후 최종 데이터 건수: {len(user_contacts):,}건")
display(user_contacts.head())
```

```python
#accounts_blockrecord

# 1. 원본 데이터 복사 및 날짜 타입/KST 변환
df_abr = blockrecord.copy()

# 💡 필수: 이전의 뺄셈 에러를 방지하기 위해 문자열을 datetime으로 변환
df_abr['created_at'] = pd.to_datetime(df_abr['created_at'])

# 💡 KST 변경: 원본 데이터가 UTC(세계 표준시)로 저장되어 있다면 9시간을 더해 한국 시간으로 맞춥니다.
# (만약 timezone 정보가 이미 포함된 데이터라면 df_abr['created_at'].dt.tz_convert('Asia/Seoul') 를 사용합니다)
df_abr['created_at'] = df_abr['created_at'] + pd.Timedelta(hours=9)

# 시간 흐름대로 정렬
df_abr = df_abr.sort_values(by=['user_id', 'reason', 'block_user_id', 'created_at'])

# 2. 유저, 차단 대상, 사유가 모두 일치하는 그룹 내에서 시간 차이(초) 계산
df_abr['time_diff_sec'] = df_abr.groupby(['user_id', 'block_user_id', 'reason'])['created_at'].diff().dt.total_seconds()

# 3. 삭제 대상 정의: 시간 차이가 60초 이내인 데이터 행들의 인덱스 추출
drop_condition = (df_abr['time_diff_sec'].notnull()) & (df_abr['time_diff_sec'] <= 60)
delete_indices = df_abr[drop_condition].index

print(f"정제 전 데이터 건수 : {len(df_abr):,}건")
print(f"삭제될 차단 중복 건수: {len(delete_indices):,}건")

# 4. 원본에서 해당 중복 데이터들만 제거 후 반영 (df_abr 기준 적용)
accounts_blockrecord = df_abr.drop(index=delete_indices)

# 5. 파생 컬럼 삭제 (메모리 정리)
accounts_blockrecord = accounts_blockrecord.drop(columns=['time_diff_sec'])

# 💡 수정: 정제 완료된 accounts_blockrecord의 길이를 출력해야 정확한 결과가 나옵니다.
print(f"🎉 정제 후 최종 데이터 건수: {len(accounts_blockrecord):,}건")
```

```python
#polls_qusetionset
import pandas as pd
import ast

# 1. 안전한 리스트 파싱 함수 (작성하신 훌륭한 코드 그대로 유지)
def parse_list_value(x):
    if pd.isna(x):
        return []
    
    if isinstance(x, list):
        return x
    
    if isinstance(x, str):
        x = x.strip()
        if x in ["", "[]", "None", "nan"]:
            return []
        try:
            parsed = ast.literal_eval(x)
            if isinstance(parsed, list):
                return parsed
            else:
                return []
        except:
            return []
    
    return []


polls_qset = polls_qset.copy()

# 💡 수정 1: 시간 변환을 데이터 분리(explode) 전에 먼저 수행합니다.
polls_qset['opening_time'] = pd.to_datetime(polls_qset['opening_time'], errors='coerce') + pd.Timedelta(hours=9)
polls_qset['created_at'] = pd.to_datetime(polls_qset['created_at'], errors='coerce') + pd.Timedelta(hours=9)

# 2. 리스트 파싱 및 질문 개수 계산
polls_qset["question_piece_id_parsed"] = polls_qset["question_piece_id_list"].apply(parse_list_value)
polls_qset["question_count"] = polls_qset["question_piece_id_parsed"].apply(len)

print("==== [세트별 질문 개수 분포] ====")
display(polls_qset["question_count"].value_counts().sort_index())

# 모든 세트가 10개인지 확인
abnormal_count = (polls_qset["question_count"] != 10).sum()
print(f"▶ 질문 개수 10개가 아닌 세트 수: {abnormal_count:,}건")

# 3. 질문 세트 리스트를 1줄씩 펼치기(explode)
polls_qset_exploded = (
    polls_qset[["id", "user_id", "status", "created_at", "opening_time", "question_piece_id_parsed"]]
    .explode("question_piece_id_parsed")
    .rename(columns={
        "id": "questionset_id",
        "question_piece_id_parsed": "question_piece_id"
    })
)

# 💡 수정 2: NaN을 허용하는 판다스의 정수 타입인 'Int64' (대문자 I)를 사용합니다.
polls_qset_exploded["question_piece_id"] = polls_qset_exploded["question_piece_id"].astype("Int64")

print("\n==== [Explode 처리 완료 데이터] ====")
display(polls_qset_exploded.head())
print(f"▶ 펼친 후 총 행 수: {len(polls_qset_exploded):,}건")
```

```python
import pandas as pd
import numpy as np

# 원본 데이터 복사 및 시간 데이터 변환
df_payments = paymenthistory.copy()
df_payments['created_at'] = pd.to_datetime(df_payments['created_at'], errors='coerce')
df_payments = df_payments.dropna(subset=['created_at'])

# ==========================================
# 1. ARPPU (유저 1인당 평균 결제 금액) 분석
# ==========================================
print("========== [1. 유저 1인당 평균 결제 금액(ARPPU) 분석] ==========")

# 💡 [필수 설정] 실제 앱스토어/구글플레이에 등록된 상품 가격(원)을 적어주세요!
# 만약 실제 가격과 다르다면 아래 숫자를 수정하시면 됩니다.
price_mapping = {
    'heart.777': 7700,
    'heart.1000': 10000,
    'heart.200': 2000,
    'heart.4000': 40000
}

# 결제 기록에 가격(price) 컬럼 추가 (매핑 안 된 상품은 임시로 1,000원으로 계산)
df_payments['price'] = df_payments['productId'].map(price_mapping).fillna(1000)

# 유저별 총 결제 금액 계산
user_total_spend = df_payments.groupby('user_id')['price'].sum()

# ARPPU 계산 (전체 결제 금액 / 결제 유저 수)
total_revenue = user_total_spend.sum()
paying_user_count = len(user_total_spend)
arppu = total_revenue / paying_user_count if paying_user_count > 0 else 0

print(f"▶ 누적 예상 총 결제액: {total_revenue:,.0f}원")
print(f"▶ 지갑을 연 유저 수: {paying_user_count:,}명")
print(f"▶ 🎯 유저 1인당 평균 결제 금액(ARPPU): {arppu:,.0f}원\n")


# ==========================================
# 2. 월별 재결제율(Cohort Retention) 분석
# ==========================================
print("========== [2. 월별 코호트 재결제율 분석] ==========")

# 1) 결제 발생 '월(Month)' 정보 추출
df_payments['order_month'] = df_payments['created_at'].dt.to_period('M')

# 2) 유저별 '최초 결제월' 구하기
first_purchase = df_payments.groupby('user_id')['order_month'].min().reset_index()
first_purchase.columns = ['user_id', 'first_month']

# 3) 원본 결제 기록에 최초 결제월 정보 합치기
df_merged = pd.merge(df_payments, first_purchase, on='user_id')

# 4) 최초 결제월로부터 몇 개월(Month Diff)이 지났는지 계산
df_merged['month_diff'] = (df_merged['order_month'] - df_merged['first_month']).apply(lambda x: x.n)

# 5) 코호트 테이블 생성 (최초 결제월 기준, N개월 후 결제한 유저 수)
cohort_data = df_merged.groupby(['first_month', 'month_diff'])['user_id'].nunique().reset_index()
cohort_pivot = cohort_data.pivot(index='first_month', columns='month_diff', values='user_id')

# 6) 유저 수를 비율(%)로 변환
cohort_sizes = cohort_pivot.iloc[:, 0] # month_diff 0 (당월 결제 유저 수)
retention_matrix = cohort_pivot.divide(cohort_sizes, axis=0) * 100

# 표를 보기 좋게 포맷팅
retention_matrix_styled = retention_matrix.style.format("{:.1f}%", na_rep="-").background_gradient(cmap='YlGn', axis=None, vmin=0, vmax=20)

# 출력
print("💡 표 읽는 법: 행(Index)은 첫 결제를 한 달, 열(Columns)은 첫 결제 후 경과한 달(0=당월, 1=다음달)을 의미합니다.")
print("예) 열 '1'의 수치가 10%라면, 해당 월에 처음 결제한 사람 중 10%가 다음 달에도 결제했다는 뜻입니다.")
display(retention_matrix_styled)
```

```python
import pandas as pd
import numpy as np

# ==========================================
# 💡 [설정] 원본 데이터 복사 및 컬럼명 확인
# ==========================================
df_users = user.copy()
df_payments = paymenthistory.copy()

# 유저 테이블의 가입일 컬럼 찾기 (일반적으로 'created_at' 또는 'date_joined')
user_time_col = 'created_at' if 'created_at' in df_users.columns else 'date_joined'

print("========== [가입 후 첫 결제까지 걸린 시간 분석] ==========")

try:
    # 1. 유저별 '최초 결제일' 구하기
    df_payments['created_at'] = pd.to_datetime(df_payments['created_at'], errors='coerce')
    first_purchase = df_payments.groupby('user_id')['created_at'].min().reset_index()
    first_purchase.columns = ['user_id', 'first_purchase_date']

    # 2. 유저 테이블에서 가입일 가져오기
    user_signup = df_users[['id', user_time_col]].copy()
    user_signup[user_time_col] = pd.to_datetime(user_signup[user_time_col], errors='coerce')
    user_signup.columns = ['user_id', 'signup_date']

    # 3. 데이터 병합 (가입일과 최초 결제일)
    df_merged = pd.merge(first_purchase, user_signup, on='user_id', how='inner')

    # 🚨 시간대(Timezone)가 다를 경우를 대비해 시간대 정보 제거 (안전한 계산을 위함)
    df_merged['first_purchase_date'] = df_merged['first_purchase_date'].dt.tz_localize(None)
    df_merged['signup_date'] = df_merged['signup_date'].dt.tz_localize(None)

    # 4. 가입 후 결제까지 걸린 '일(Days)' 계산
    df_merged['days_to_purchase'] = (df_merged['first_purchase_date'] - df_merged['signup_date']).dt.days

    # 데이터 오류 방지 (결제일이 가입일보다 과거인 이상 데이터 0으로 처리)
    df_merged['days_to_purchase'] = np.where(df_merged['days_to_purchase'] < 0, 0, df_merged['days_to_purchase'])

    # 5. 기간별 그룹화 (구간 나누기)
    bins = [-1, 0, 1, 3, 7, 14, 30, 99999]
    labels = ['당일 결제 (0일)', '1일 후', '2~3일 후', '4~7일 후', '8~14일 후', '15~30일 후', '31일 이상']
    
    df_merged['purchase_timing'] = pd.cut(df_merged['days_to_purchase'], bins=bins, labels=labels)

    # 6. 통계 집계 및 비율 계산
    timing_stats = df_merged['purchase_timing'].value_counts().reset_index()
    timing_stats.columns = ['결제 소요 기간', '유저 수']
    
    # 정렬 (라벨 순서대로)
    timing_stats['결제 소요 기간'] = pd.Categorical(timing_stats['결제 소요 기간'], categories=labels, ordered=True)
    timing_stats = timing_stats.sort_values('결제 소요 기간')

    # 비율 계산
    total_payers = timing_stats['유저 수'].sum()
    timing_stats['비율(%)'] = (timing_stats['유저 수'] / total_payers * 100).round(1)

    print(f"▶ 결제 분석 대상 유저 수: {total_payers:,}명")
    display(timing_stats.style.bar(subset=['비율(%)'], color='#ff9999', vmin=0, vmax=100))

except KeyError as e:
    print(f"⚠️ 컬럼을 찾을 수 없습니다: {e}")
    print("유저 테이블의 가입일 컬럼명이나 결제 테이블의 유저 ID 컬럼명이 맞는지 확인해 주세요.")
except Exception as e:
    print(f"⚠️ 알 수 없는 에러 발생: {e}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, classification_report

# 시각화를 위한 한글 폰트 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

print("========== [4. 머신러닝 기반 재결제 예측(Supervised Learning)] ==========")

try:
    # ==========================================
    # 0. 데이터 전처리 및 필수 변수(Feature) 준비
    # ==========================================
    df_pay = paymenthistory.copy()
    df_evt = hackle_events.copy()
    h_prop = hackle_prop.drop_duplicates(subset=['session_id']).copy()

    # ID 문자열 통일
    df_pay['user_id'] = df_pay['user_id'].astype(str)
    df_evt['session_id'] = df_evt['session_id'].astype(str)
    h_prop['session_id'] = h_prop['session_id'].astype(str)

    # 이벤트 테이블에 user_id 병합
    if 'user_id' not in df_evt.columns:
        df_evt = pd.merge(df_evt, h_prop[['session_id', 'user_id']], on='session_id', how='left')
    df_evt['user_id'] = df_evt['user_id'].astype(str)

    # 🚨 NameError를 해결하기 위해 user_pay_counts를 여기서 직접 생성합니다.
    user_pay_counts = df_pay.groupby('user_id').size().reset_index(name='pay_count')


    # ==========================================
    # 1. Feature Engineering (모델 학습용 데이터셋 구축)
    # ==========================================
    # 대상: 최소 1회 이상 결제한 유저
    ml_users = user_pay_counts.copy()

    # Target 설정: 2회 이상 결제면 1(재결제), 1회면 0(이탈)
    ml_users['target_retained'] = (ml_users['pay_count'] >= 2).astype(int)

    # Feature 생성: 유저별 행동(event_key) 발생 횟수를 피벗하여 변수로 사용
    evt_features = df_evt[df_evt['user_id'].isin(ml_users['user_id'])]
    user_event_pivot = pd.crosstab(evt_features['user_id'], evt_features['event_key'])

    # 최종 데이터셋 결합
    ml_data = pd.merge(ml_users[['user_id', 'target_retained']], user_event_pivot, on='user_id', how='left')


    # ==========================================
    # 2. X (독립 변수)와 y (종속 변수) 분리 및 전처리
    # ==========================================
    X = ml_data.drop(columns=['user_id', 'target_retained'])
    y = ml_data['target_retained']

    # 결측치 처리 (SimpleImputer 사용)
    imputer = SimpleImputer(strategy='constant', fill_value=0)
    X_imputed = pd.DataFrame(imputer.fit_transform(X), columns=X.columns)

    # 학습용/테스트용 데이터 분할 (8:2)
    X_train, X_test, y_train, y_test = train_test_split(X_imputed, y, test_size=0.2, random_state=42)


    # ==========================================
    # 3. 모델 학습 (Random Forest Classifier)
    # ==========================================
    rf_model = RandomForestClassifier(n_estimators=100, random_state=42, max_depth=5)
    rf_model.fit(X_train, y_train)


    # ==========================================
    # 4. 모델 평가
    # ==========================================
    y_pred = rf_model.predict(X_test)
    print(f"▶ 예측 정확도(Accuracy): {accuracy_score(y_test, y_pred):.2f}")
    print("\n[상세 분류 리포트]")
    print(classification_report(y_test, y_pred, target_names=['이탈(0)', '재결제(1)']))


    # ==========================================
    # 5. Feature Importance 시각화 (최신 seaborn 문법 적용)
    # ==========================================
    importance = pd.DataFrame({
        'Feature (행동)': X.columns,
        'Importance': rf_model.feature_importances_
    }).sort_values(by='Importance', ascending=False)

    print("\n========== [머신러닝이 찾아낸 재결제 핵심 행동 Top 5] ==========")
    display(importance.head(5).style.bar(subset=['Importance'], color='#86bf91'))

    plt.figure(figsize=(10, 5))
    
    sns.barplot(
        x='Importance', 
        y='Feature (행동)', 
        data=importance.head(10), 
        hue='Feature (행동)',
        palette='viridis', 
        legend=False
    )
    plt.title('재결제 여부를 결정짓는 주요 행동 지표 (Feature Importance)')
    plt.xlabel('영향도 (가중치)')
    plt.ylabel('')
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"⚠️ 에러가 발생했습니다: {e}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 시각화 폰트 및 마이너스 기호 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

print("========== [유저 생애 가치(LTV) 추정 분석] ==========")

try:
    # ==========================================
    # 1. 데이터 준비 및 가격 매핑
    # ==========================================
    df_pay = paymenthistory.copy()
    df_pay['created_at'] = pd.to_datetime(df_pay['created_at'], errors='coerce')
    df_pay = df_pay.dropna(subset=['created_at'])
    
    # 상품별 실제 가격(원) 매핑 (실제 가격에 맞게 수정 가능)
    price_mapping = {
        'heart.777': 7700,
        'heart.1000': 10000,
        'heart.200': 2000,
        'heart.4000': 40000
    }
    df_pay['price'] = df_pay['productId'].map(price_mapping).fillna(1000)
    
    # 결제 발생 '월(Month)' 정보 추출
    df_pay['order_month'] = df_pay['created_at'].dt.to_period('M')

    # ==========================================
    # 2. 코호트(최초 결제월) 설정 및 경과 기간 계산
    # ==========================================
    # 유저별 최초 결제월 추출
    first_purchase = df_pay.groupby('user_id')['order_month'].min().reset_index()
    first_purchase.columns = ['user_id', 'first_month']
    
    # 원본에 병합 후, 최초 결제월로부터 몇 달이 지났는지(month_diff) 계산
    df_ltv = pd.merge(df_pay, first_purchase, on='user_id')
    df_ltv['month_diff'] = (df_ltv['order_month'] - df_ltv['first_month']).apply(lambda x: x.n)

    # ==========================================
    # 3. LTV(누적 ARPU) 계산
    # ==========================================
    # 1) 코호트별 & 경과월별 총 결제 금액 집계
    cohort_revenue = df_ltv.groupby(['first_month', 'month_diff'])['price'].sum().reset_index()
    
    # 2) 코호트별 최초 결제 유저 수 집계
    cohort_users = first_purchase.groupby('first_month')['user_id'].nunique().reset_index()
    cohort_users.columns = ['first_month', 'total_users']
    
    # 3) 병합 후 월별 매출을 '누적 매출(Cumulative Revenue)'로 변환
    cohort_data = pd.merge(cohort_revenue, cohort_users, on='first_month')
    cohort_data['cumulative_revenue'] = cohort_data.groupby('first_month')['price'].cumsum()
    
    # 4) LTV 산출 = 누적 매출 / 해당 코호트의 전체 유저 수
    cohort_data['LTV'] = cohort_data['cumulative_revenue'] / cohort_data['total_users']

    # ==========================================
    # 4. LTV 테이블 (Pivot) 생성 및 출력
    # ==========================================
    ltv_pivot = cohort_data.pivot(index='first_month', columns='month_diff', values='LTV')
    
    print("\n💡 [월별 코호트 누적 LTV 테이블 (단위: 원)]")
    print("행(Index)은 첫 결제월, 열(Columns)은 첫 결제 후 경과한 달(0=당월)을 의미합니다.")
    display(ltv_pivot.style.format("{:,.0f}", na_rep="-").background_gradient(cmap='Greens', axis=1))

    # ==========================================
    # 5. LTV 성장 곡선 (Growth Curve) 시각화
    # ==========================================
    plt.figure(figsize=(12, 6))
    
    # 기간이 너무 짧은 코호트는 제외하고, 데이터가 충분한 주요 코호트만 선별하여 그리기
    for cohort in ltv_pivot.index:
        # 데이터가 존재하는 month_diff 구간만 추출
        cohort_series = ltv_pivot.loc[cohort].dropna()
        if len(cohort_series) > 1:  # 최소 2개월 이상 데이터가 있는 코호트만 플로팅
            plt.plot(cohort_series.index, cohort_series.values, marker='o', label=str(cohort), linewidth=2)

    plt.title('월별 코호트 유저 생애 가치(LTV) 성장 곡선', fontsize=15, pad=15)
    plt.xlabel('첫 결제 후 경과 개월 수 (Month)', fontsize=12)
    plt.ylabel('1인당 누적 결제액 (LTV, 원)', fontsize=12)
    
    # x축 눈금 정수 처리
    plt.xticks(range(int(cohort_data['month_diff'].max()) + 1))
    
    plt.legend(title='첫 결제월 (Cohort)', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"⚠️ 분석 중 에러가 발생했습니다: {e}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 시각화를 위한 한글 폰트 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

print("========== [1. K-Means 기반 유저 행동 군집화] ==========")

try:
    # ==========================================
    # 0. 데이터 준비 (이전 분석에서 집계한 유저별 피벗 테이블 활용)
    # ==========================================
    # 실무에서는 앞서 만드신 user_event_pivot 데이터 등에 '결제액', '접속일수' 등을 병합하여 사용합니다.
    # 여기서는 군집화 로직을 명확히 보여드리기 위해 가상의 집계 데이터를 구성한다고 가정합니다.
    
    # 예시 변수: user_id, 주간_접속_횟수, 누적_투표_수, 질문_열기_횟수, 하트_소모량
    features = ['주간_접속_횟수', '누적_투표_수', '질문_열기_횟수', '하트_소모량']
    
    # df_user_features = ... (여기에 실제 병합하신 데이터프레임을 넣으시면 됩니다)
    # X_raw = df_user_features[features]
    
    # [코드 실행을 위한 임시 데이터 생성 (실제 적용 시 삭제)]
    np.random.seed(42)
    X_raw = pd.DataFrame({
        '주간_접속_횟수': np.random.poisson(3, 1000),
        '누적_투표_수': np.random.poisson(10, 1000),
        '질문_열기_횟수': np.random.poisson(2, 1000),
        '하트_소모량': np.random.exponential(500, 1000)
    })
    
    # ==========================================
    # 1. 데이터 스케일링 (표준화)
    # ==========================================
    # K-Means는 '거리' 기반 알고리즘이므로, 단위가 큰 '하트_소모량'이 결과를 지배하지 않도록 
    # 모든 변수의 단위를 평균 0, 분산 1로 맞춰줍니다.
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_raw)

    # ==========================================
    # 2. K-Means 모델 학습
    # ==========================================
    n_clusters = 4  # 나눌 그룹의 개수 (보통 3~5개가 적당합니다)
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(X_scaled)
    
    # 원본 데이터에 군집(Cluster) 라벨 추가
    X_raw['Cluster'] = cluster_labels

    # ==========================================
    # 3. 군집별 특징(프로파일링) 분석 및 시각화
    # ==========================================
    # 군집별 각 변수의 평균값 계산
    cluster_profile = X_raw.groupby('Cluster').mean().round(1)
    
    # 군집별 유저 수 계산
    cluster_profile['유저_수(명)'] = X_raw.groupby('Cluster').size()
    
    print("\n💡 [머신러닝이 분류한 유저 군집별 평균 행동 지표]")
    display(cluster_profile.style.background_gradient(cmap='Blues'))

    # 히트맵을 통한 직관적인 군집 특징 시각화 (유저 수 제외하고 스케일링된 값으로 표현)
    centroids_scaled = pd.DataFrame(kmeans.cluster_centers_, columns=features)
    
    plt.figure(figsize=(10, 6))
    sns.heatmap(centroids_scaled, cmap='coolwarm', annot=True, fmt=".2f", center=0)
    plt.title('각 군집(Cluster)별 행동 변수 중요도 (0 기준 양수면 활발, 음수면 저조)', fontsize=14, pad=15)
    plt.xlabel('행동 변수 (Features)', fontsize=12)
    plt.ylabel('군집 (Cluster)', fontsize=12)
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"⚠️ 에러가 발생했습니다: {e}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

print("\n========== [3. 유저 라이프사이클 단계별 분류] ==========")

try:
    # ==========================================
    # 1. 기준 일자(Today) 및 데이터 준비
    # ==========================================
    # 분석 시점의 기준 날짜 (통상적으로 데이터의 가장 마지막 날짜를 사용합니다)
    # 여기서는 예시로 2024년 5월 1일로 가정합니다. (실무: df_evt['event_datetime'].max())
    analysis_date = pd.to_datetime('2024-05-01')
    
    # [코드 실행을 위한 임시 데이터 생성 (실제 적용 시 삭제)]
    # user_id, 가입일(signup_date), 마지막_접속일(last_login_date)이 필요합니다.
    user_lifecycle = pd.DataFrame({
        'user_id': range(1, 1001),
        'signup_date': pd.date_range(start='2024-01-01', end='2024-05-01', periods=1000),
        'last_login_date': pd.date_range(start='2024-03-01', end='2024-05-01', periods=1000)
    })
    
    # 랜덤하게 과거 날짜 섞기
    user_lifecycle['last_login_date'] = user_lifecycle['last_login_date'].sample(frac=1).reset_index(drop=True)

    # ==========================================
    # 2. 경과 일수(Days Since) 계산
    # ==========================================
    user_lifecycle['days_since_signup'] = (analysis_date - user_lifecycle['signup_date']).dt.days
    user_lifecycle['days_since_last_login'] = (analysis_date - user_lifecycle['last_login_date']).dt.days

    # ==========================================
    # 3. 룰(Rule) 기반 라이프사이클 상태 부여
    # ==========================================
    # 다중 조건 설정
    conditions = [
        (user_lifecycle['days_since_signup'] <= 7),                                      # 신규 (가입 7일 이내)
        (user_lifecycle['days_since_signup'] > 7) & (user_lifecycle['days_since_last_login'] <= 14), # 활성 (신규가 아니며 최근 14일 내 접속)
        (user_lifecycle['days_since_last_login'] > 14) & (user_lifecycle['days_since_last_login'] <= 30), # 이탈 위기 (15~30일 미접속)
        (user_lifecycle['days_since_last_login'] > 30)                                   # 이탈 (30일 초과 미접속)
    ]
    
    # 부여할 상태(Label)
    choices = ['1. 신규 (New)', '2. 활성 (Active)', '3. 이탈 위기 (At-Risk)', '4. 이탈 (Churned)']
    
    # np.select를 사용하여 조건에 맞는 상태 매핑
    user_lifecycle['lifecycle_stage'] = np.select(conditions, choices, default='기타')

    # ==========================================
    # 4. 현황 집계 및 시각화
    # ==========================================
    stage_counts = user_lifecycle['lifecycle_stage'].value_counts().reset_index()
    stage_counts.columns = ['라이프사이클 단계', '유저 수(명)']
    stage_counts = stage_counts.sort_values(by='라이프사이클 단계')
    
    stage_counts['비율(%)'] = (stage_counts['유저 수(명)'] / stage_counts['유저 수(명)'].sum() * 100).round(1)
    
    print("\n💡 [유저 라이프사이클 현황 집계표]")
    display(stage_counts.style.bar(subset=['비율(%)'], color='#ffb3b3', vmin=0, vmax=100))

    # 도넛 차트(Donut Chart) 시각화
    plt.figure(figsize=(8, 8))
    colors = ['#8fd9b6', '#ff9999', '#ffcc99', '#c2c2f0']
    
    wedges, texts, autotexts = plt.pie(
        stage_counts['유저 수(명)'], 
        labels=stage_counts['라이프사이클 단계'], 
        autopct='%1.1f%%', 
        startangle=90, 
        colors=colors,
        wedgeprops={'width': 0.4, 'edgecolor': 'w'}  # width 속성이 도넛 모양을 만듭니다.
    )
    
    plt.title('전체 유저 라이프사이클 비중', fontsize=16, pad=20)
    plt.setp(autotexts, size=11, weight="bold")
    plt.tight_layout()
    plt.show()
    
    print("\n🚀 [실무 활용 팁]")
    print("- 신규(New) 단계 유저에게는 온보딩 푸시를, 이탈 위기(At-Risk) 유저에게는 하트 지급 복귀 알림을 보내는 방식(CRM)으로 연계하세요!")

except Exception as e:
    print(f"⚠️ 에러가 발생했습니다: {e}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 시각화를 위한 한글 폰트 및 마이너스 기호 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

print("========== [유저 유형화: 소셜 생태계 역할(Social Role) 분류] ==========")

try:
    # ==========================================
    # 1. 데이터 준비 (가상 데이터 생성)
    # ==========================================
    # 실무에서는 투표 로그 테이블에서 유저별 '보낸 투표 수'와 '받은 투표 수'를 집계하여 사용합니다.
    np.random.seed(42)
    
    # votes_given: 내가 남에게 한 투표 수
    # votes_received: 내가 남으로부터 받은 투표 수
    df_social = pd.DataFrame({
        'user_id': range(1, 1001),
        'votes_given': np.random.lognormal(mean=2.5, sigma=1.2, size=1000).astype(int),
        'votes_received': np.random.lognormal(mean=2.0, sigma=1.5, size=1000).astype(int)
    })
    
    # ==========================================
    # 2. 역할 분류를 위한 상위 기준선(Threshold) 설정
    # ==========================================
    # 소셜 네트워크에서는 상위 10~20%가 생태계를 주도하므로, 80백분위수(Top 20%)를 기준으로 삼습니다.
    given_threshold = df_social['votes_given'].quantile(0.80)
    received_threshold = df_social['votes_received'].quantile(0.80)

    # ==========================================
    # 3. 룰(Rule) 기반 소셜 페르소나 매핑
    # ==========================================
    conditions = [
        # 투표도 많이 하고, 받기도 많이 받는 유저 (앱의 얼굴)
        (df_social['votes_given'] >= given_threshold) & (df_social['votes_received'] >= received_threshold),
        
        # 받지는 못하지만, 남들에게 투표를 엄청나게 뿌리는 유저 (생태계 엔진)
        (df_social['votes_given'] >= given_threshold) & (df_social['votes_received'] < received_threshold),
        
        # 투표는 안 하면서, 받기는 엄청나게 받는 유저 (인플루언서/눈팅족)
        (df_social['votes_given'] < given_threshold) & (df_social['votes_received'] >= received_threshold),
        
        # 투표도 안 하고, 받지도 못하는 유저 (소외된 유저)
        (df_social['votes_given'] < given_threshold) & (df_social['votes_received'] < received_threshold)
    ]
    
    choices = [
        '1. 슈퍼스타 (핵심 교류자)', 
        '2. 기버 (투표 공급자)', 
        '3. 인플루언서 (수신 특화)', 
        '4. 관전자/소외자 (저관여)'
    ]
    
    df_social['social_persona'] = np.select(conditions, choices, default='기타')

    # ==========================================
    # 4. 페르소나별 현황 집계 및 시각화
    # ==========================================
    # 1) 페르소나별 유저 수 집계
    persona_stats = df_social['social_persona'].value_counts().reset_index()
    persona_stats.columns = ['소셜 페르소나', '유저 수(명)']
    persona_stats = persona_stats.sort_values('소셜 페르소나')
    persona_stats['비율(%)'] = (persona_stats['유저 수(명)'] / len(df_social) * 100).round(1)
    
    print("\n💡 [소셜 페르소나별 분포 현황]")
    display(persona_stats.style.bar(subset=['비율(%)'], color='#d9b3ff', vmin=0, vmax=100))

    # 2) 산점도(Scatter Plot)를 통한 생태계 시각화
    plt.figure(figsize=(10, 8))
    
    # 시각적 깔끔함을 위해 상위 95%까지만 축 제한
    x_max = df_social['votes_given'].quantile(0.95)
    y_max = df_social['votes_received'].quantile(0.95)
    
    sns.scatterplot(
        data=df_social, 
        x='votes_given', 
        y='votes_received', 
        hue='social_persona', 
        palette='husl', 
        alpha=0.6,
        s=50
    )
    
    # 기준선(Top 20%) 그리기
    plt.axvline(given_threshold, color='gray', linestyle='--', label=f'발신 Top 20% ({int(given_threshold)}회)')
    plt.axhline(received_threshold, color='gray', linestyle=':', label=f'수신 Top 20% ({int(received_threshold)}회)')
    
    plt.xlim(-5, x_max)
    plt.ylim(-5, y_max)
    
    # 이모지를 제외한 깔끔한 텍스트 제목 사용
    plt.title('보낸 투표 vs 받은 투표 기반 소셜 페르소나 매트릭스', fontsize=16, pad=15)
    plt.xlabel('보낸 투표 수 (Give)', fontsize=12)
    plt.ylabel('받은 투표 수 (Receive)', fontsize=12)
    
    plt.legend(title='분류 기준', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

    print("\n🚀 [페르소나별 맞춤형 액션 플랜]")
    print("- 슈퍼스타 & 기버: 이들이 투표를 멈추면 앱이 망합니다. '이달의 프로 투표러' 같은 칭호나 무제한 하트를 몰래 지급해서 무조건 잡아둬야 합니다.")
    print("- 인플루언서: 본인이 받은 투표의 '초성'을 까보고 싶어 안달 난 유저들입니다. 과금(결제)을 가장 강하게 유도해야 하는 최우선 타겟입니다.")
    print("- 관전자/소외자: 아무도 이들에게 투표하지 않아서 재미를 못 느끼는 상태입니다. 운영자 봇(Bot)을 활용해 위로성 투표를 보내 앱 체류 시간을 늘려야 합니다.")

except Exception as e:
    print(f"⚠️ 에러가 발생했습니다: {e}")
```
